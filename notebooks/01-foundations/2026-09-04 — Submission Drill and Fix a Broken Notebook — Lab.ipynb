### How to read an error before you fix it

Python's traceback is read **bottom up**. The last line names the exception and the problem; the line above it points at your code.

```
Traceback (most recent call last):
  File "<stdin>", line 2, in <module>
    print(greeting)
NameError: name 'greeting' is not defined
```

Four exceptions cover most of what you will hit this semester:

| Exception | Usually means |
|---|---|
| `NameError` | The variable was never created, or the cell that created it never ran |
| `TypeError` | You mixed types — text where a number belongs is the classic |
| `KeyError` | That column or dictionary key does not exist. Check spelling and case |
| `FileNotFoundError` | The path is wrong for *this* environment |

### Stretch

Write a `check_orders(df)` function that raises a clear error if the frame is missing either expected column, has zero rows, or contains a negative value in `units`. Then call it on `orders`. This is the shape of validation you will want in both projects.

````python
import pandas as pd
from io import StringIO

def check_orders(df):
    """
    Validate that `df` looks like a proper orders frame before using it.
    Raises a clear, specific error the moment something is wrong.
    """
    # Adjusting expected_columns to match the actual schema of the 'orders' DataFrame
    expected_columns = {'game', 'units'}
    missing = expected_columns - set(df.columns)
    if missing:
        raise ValueError(f"orders is missing expected column(s): {sorted(missing)}")

    if len(df) == 0:
        raise ValueError("orders has zero rows")

    # Check for 'units' column existence before attempting to access it for negative values
    if 'units' in df.columns and (df['units'] < 0).any():
        bad_rows = df.index[df['units'] < 0].tolist()
        raise ValueError(f"orders contains negative 'units' at row(s): {bad_rows}")

    print("orders passed all checks:", len(df), "rows")

# Re-create the 'orders' DataFrame specifically for this check,
# as the 'orders' variable might have been overwritten as a list by a previous cell.
# The 'SAMPLE' variable is available from cell H8T0KDW-z2BI. Re-defining it here for clarity within the cell.
SAMPLE = '''game,units\nNC State,1450\nLouisville,1610\n'''
orders_df = pd.read_csv(StringIO(SAMPLE))

check_orders(orders_df)
````
